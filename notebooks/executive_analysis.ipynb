# LATAM Market Expansion Intelligence — 2025

Five indicators, ten countries, and a like-for-like 2024 comparison. Source snapshot: 21 September 2026; analysis updated: 30 September 2026. Run `python src/build_analysis.py` from the repository root before opening this notebook.


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Image
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = ROOT / 'data/processed'
ranking = pd.read_csv(DATA / 'market_screening_2025.csv')
comparison = pd.read_csv(DATA / 'market_screening_comparison_2024_2025.csv')
sensitivity = pd.read_csv(DATA / 'weight_sensitivity_2025.csv')
coverage = pd.read_csv(DATA / 'data_coverage_2024_2025.csv')
metadata = json.loads((DATA / 'analysis_metadata.json').read_text())
print('Reference year:', metadata['reference_year'])
print('Source acquired:', ', '.join(metadata['source_retrieval_dates']))
```

## 2025 ranking
Brazil, Mexico and Costa Rica lead the core-five model. Scores do not measure commercial success or market-entry risk.


```python
display(ranking[['rank','country','screening_score']].round(2))
display(Image(filename=str(ROOT / 'reports/figures/market_screening_2025.png')))
```

## Coverage audit
The score uses the same five complete indicators for all countries. Missing internet use, inflation and trade observations are not imputed.
